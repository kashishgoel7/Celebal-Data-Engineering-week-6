```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Week6").getOrCreate()
```

```python
data = [(1, "Widget A", "Electronics", 250.0),
        (2, "Widget B", "Furniture", 180.0),
        (3, "Widget C", "Electronics", 320.0)]

columns = ["product_id", "name", "category", "price"]

df = spark.createDataFrame(data, columns)
df.show()
```

```python
df.printSchema()
```

```python
from pyspark.sql.types import DoubleType

df = df.withColumnRenamed("name", "product_name")
df = df.withColumn("price", df["price"].cast(DoubleType()))
df.show()
```

```python
df.filter(df.category == "Electronics").show()
```

```python
df = df.withColumn("final_price", df["price"] * 1.18)
df.show()
```

```python
df.groupBy("category").count().show()
```

```python
df.write.mode("overwrite").csv("output_csv", header=True)
```

```python
df.write.mode("overwrite").parquet("output_parquet")
```

```python
orders_data = [
    (1, "Completed", 2500),
    (2, "Pending", 500),
    (3, "Completed", 1500),
    (4, "Completed", 800)
]

orders_columns = ["order_id", "status", "amount"]

df_orders = spark.createDataFrame(orders_data, orders_columns)

df_orders.filter(
    (df_orders.status == "Completed") &
    (df_orders.amount > 1000)
).show()
```

```python
region_data = [
    ("North", "Low"),
    ("South", "High"),
    ("East", "Medium"),
    ("North", "High")
]

region_columns = ["region", "priority"]

df_region = spark.createDataFrame(region_data, region_columns)

df_region.filter(
    (df_region.region == "North") |
    (df_region.priority == "High")
).show()
```
